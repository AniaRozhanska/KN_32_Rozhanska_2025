```python
from abc import ABC, abstractmethod
from typing import List


class Person(ABC):
    """Абстрактний клас для представлення особи"""
    
    def __init__(self, name: str, age: int):
        """
        Ініціалізує об'єкт Person
        
        Args:
            name (str): Ім'я особи
            age (int): Вік особи
        """
        self.name = name
        self.age = age
    
    @abstractmethod
    def get_role(self) -> str:
        """Абстрактний метод для отримання ролі особи"""
        pass


class Student(Person):
    """Клас для представлення студента"""
    
    def __init__(self, name: str, age: int):
        """
        Ініціалізує об'єкт Student
        
        Args:
            name (str): Ім'я студента
            age (int): Вік студента
        """
        super().__init__(name, age)
        self.__grades: List[float] = []  # Приватний список оцінок (інкапсуляція)
    
    def add_grade(self, grade: float) -> None:
        """
        Додає оцінку до списку оцінок студента
        
        Args:
            grade (float): Оцінка для додавання (від 0 до 100)
        """
        if not (0 <= grade <= 100):
            raise ValueError("Оцінка повинна бути від 0 до 100")
        self.__grades.append(grade)
    
    def average(self) -> float:
        """
        Обчислює середнє арифметичне всіх оцінок студента
        
        Returns:
            float: Середня оцінка або 0, якщо оцінок немає
        """
        if not self.__grades:
            return 0.0
        return sum(self.__grades) / len(self.__grades)
    
    def min_grade(self) -> float:
        """
        Знаходить мінімальну оцінку студента
        
        Returns:
            float: Мінімальна оцінка або None, якщо оцінок немає
        """
        if not self.__grades:
            return None
        return min(self.__grades)
    
    def max_grade(self) -> float:
        """
        Знаходить максимальну оцінку студента
        
        Returns:
            float: Максимальна оцінка або None, якщо оцінок немає
        """
        if not self.__grades:
            return None
        return max(self.__grades)
    
    def get_role(self) -> str:
        """
        Повертає роль особи
        
        Returns:
            str: 'Student'
        """
        return "Student"
    
    def get_grades(self) -> List[float]:
        """
        Повертає список всіх оцінок (для читання)
        
        Returns:
            List[float]: Список оцінок
        """
        return self.__grades.copy()


class Teacher(Person):
    """Клас для представлення вчителя"""
    
    def __init__(self, name: str, age: int, subject: str):
        """
        Ініціалізує об'єкт Teacher
        
        Args:
            name (str): Ім'я вчителя
            age (int): Вік вчителя
            subject (str): Предмет, який викладає вчитель
        """
        super().__init__(name, age)
        self.subject = subject
    
    def get_role(self) -> str:
        """
        Повертає роль особи
        
        Returns:
            str: 'Teacher'
        """
        return "Teacher"
    
    def evaluate(self, student: Student, grade: float) -> None:
        """
        Оцінює студента і додає оцінку до його списку
        
        Args:
            student (Student): Об'єкт студента
            grade (float): Оцінка для додавання (від 0 до 100)
        """
        print(f"{self.name} оцінює студента {student.name} на {grade} балів з предмету {self.subject}")
        student.add_grade(grade)


# ============== ПРИКЛАД ВИКОРИСТАННЯ ==============

# Створення вчителя
teacher = Teacher("Іван Петренко", 35, "Математика")

# Створення студентів
student1 = Student("Марія Коваленко", 20)
student2 = Student("Петро Сидоренко", 21)

print("ОЦІНЮВАННЯ СТУДЕНТІВ:")
print("="*60)

# Вчитель оцінює студентів
teacher.evaluate(student1, 85)
teacher.evaluate(student1, 90)
teacher.evaluate(student1, 78)

teacher.evaluate(student2, 92)
teacher.evaluate(student2, 88)
teacher.evaluate(student2, 95)

# Виведення інформації про студентів
print("\n" + "="*60)
print(f"Студент: {student1.name} (вік: {student1.age})")
print(f"Роль: {student1.get_role()}")
print(f"Оцінки: {student1.get_grades()}")
print(f"Середня оцінка: {student1.average():.2f}")
print(f"Мінімальна оцінка: {student1.min_grade()}")
print(f"Максимальна оцінка: {student1.max_grade()}")

print("\n" + "="*60)
print(f"Студент: {student2.name} (вік: {student2.age})")
print(f"Роль: {student2.get_role()}")
print(f"Оцінки: {student2.get_grades()}")
print(f"Середня оцінка: {student2.average():.2f}")
print(f"Мінімальна оцінка: {student2.min_grade()}")
print(f"Максимальна оцінка: {student2.max_grade()}")

print("\n" + "="*60)
print(f"Вчитель: {teacher.name} (вік: {teacher.age})")
print(f"Роль: {teacher.get_role()}")
print(f"Предмет: {teacher.subject}")
```
